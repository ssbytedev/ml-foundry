## Task 5.1: Softmax

Softmax converts logits into probabilities:

$$
\text{softmax}(z)_i = \frac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}}
$$

### Stable Form

To improve numerical stability, subtract the maximum logit from each row before exponentiation:

$$
\text{softmax}(z)_i =
\frac{e^{z_i - \max(z)}}{\sum_{j=1}^{K} e^{z_j - \max(z)}}
$$

This prevents overflow when logits contain very large values.
